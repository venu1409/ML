```python
#exp 17
import pandas as pd

# Read the dataset
df = pd.read_csv("winequality-red.csv", sep=";")

print("Original Dataset:")
print(df.head(10))

# Separate input features and target
X = df.drop("quality", axis=1)
y = df["quality"]


# ==================================================
# 1. MIN-MAX NORMALIZATION
# ==================================================

minmax = pd.DataFrame()

for col in X.columns:
    minimum = X[col].min()
    maximum = X[col].max()

    minmax[col] = (X[col] - minimum) / (maximum - minimum)

# Add target column
minmax["quality"] = y

print("\nMin-Max Normalized Data:")
print(minmax.head(10))


# ==================================================
# 2. Z-SCORE STANDARDIZATION
# ==================================================

zscore = pd.DataFrame()

for col in X.columns:
    mean = X[col].mean()
    std = X[col].std()

    zscore[col] = (X[col] - mean) / std

# Add target column
zscore["quality"] = y

print("\nZ-Score Standardized Data:")
print(zscore.head(10))


# ==================================================
# Save the results
# ==================================================

minmax.to_csv("winequality_red_minmax.csv", index=False)
zscore.to_csv("winequality_red_zscore.csv", index=False)

print("\nFiles saved successfully!")
print("1. winequality_red_minmax.csv")
print("2. winequality_red_zscore.csv")
```
