```python
#exp 18
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler

# ------------------------------------------
# Read the dataset
# ------------------------------------------

df = pd.read_csv("winequality-red.csv", sep=";")

print("Original Dataset:")
print(df.head(10))


# ------------------------------------------
# Separate features and target
# ------------------------------------------

X = df.drop("quality", axis=1)
y = df["quality"]


# ==========================================
# 1. MIN-MAX NORMALIZATION
# ==========================================

minmax_scaler = MinMaxScaler()

X_minmax = minmax_scaler.fit_transform(X)

minmax_df = pd.DataFrame(
    X_minmax,
    columns=X.columns
)

# Add target column
minmax_df["quality"] = y.values

print("\n====================================")
print("MIN-MAX NORMALIZED DATA")
print("====================================")

print(minmax_df.head(10))


# ==========================================
# 2. Z-SCORE STANDARDIZATION
# ==========================================

standard_scaler = StandardScaler()

X_zscore = standard_scaler.fit_transform(X)

zscore_df = pd.DataFrame(
    X_zscore,
    columns=X.columns
)

# Add target column
zscore_df["quality"] = y.values

print("\n====================================")
print("Z-SCORE STANDARDIZED DATA")
print("====================================")

print(zscore_df.head(10))


# ==========================================
# Save the results
# ==========================================

minmax_df.to_csv(
    "winequality_red_minmax1.csv",
    index=False
)

zscore_df.to_csv(
    "winequality_red_zscore1.csv",
    index=False
)

print("\nFiles saved successfully!")
print("1. winequality_red_minmax1.csv")
print("2. winequality_red_zscore1.csv")
```

```python
#experiment 19
import pandas as pd

# Read dataset
df = pd.read_csv("winequality-red.csv", sep=";")

print("Original Dataset:")
print(df.head(10))

# Separate features and target
X = df.drop("quality", axis=1)
y = df["quality"]

# Min-Max Scaling: Range 3 to 5
new_min = 3
new_max = 5

X_scaled = (X - X.min()) / (X.max() - X.min())
X_scaled = X_scaled * (new_max - new_min) + new_min

# Convert to DataFrame
scaled_df = pd.DataFrame(X_scaled, columns=X.columns)

# Add target column
scaled_df["quality"] = y.values

print("\n====================================")
print("MIN-MAX SCALED DATA")
print("Range: 3 to 5")
print("====================================")

print(scaled_df.head(10))

# Save result
scaled_df.to_csv("winequality_red_minmax.csv", index=False)

print("\nFile saved successfully!")
print("winequality_red_minmax.csv")
```
